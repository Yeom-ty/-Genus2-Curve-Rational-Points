```python
from sympy import symbols, solve, Poly, QQ
x, y = symbols('x y')

curve = Poly(y**2 - (x**5 - 2*x + 3), x, y)

rational_points = []
for x_val in range(-5, 6):
    eq = curve.subs(x, x_val)
    y_solutions = solve(eq, y, domain=QQ)
    for y_sol in y_solutions:
        rational_points.append((x_val, y_sol))

print( rational_points)
```

```python
from sympy import I
import matplotlib.pyplot as plt


real_points = []
for point in rational_points:

    if I not in point[1].atoms():
        real_points.append((point[0], float(point[1])))


if real_points:
    x_coords = [p[0] for p in real_points]
    y_coords = [p[1] for p in real_points]
    plt.figure(figsize=(8, 6))
    plt.scatter(x_coords, y_coords, color='darkblue', s=80)
    plt.title('Real Rational Points on Algebraic Curve')
    plt.xlabel('x')
    plt.ylabel('y')
    plt.grid(alpha=0.5)
    plt.savefig('/content/real_rational_points.png')
else:
    print("当前曲线在x∈[-5,5]范围内无实数有理点")
```
